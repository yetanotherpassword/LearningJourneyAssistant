```python
import pandas as pd

# Any export produced by `python -m lja.export --out <dir>`. This path is the
# default Excel run from python/; repoint it at a Moodle or --anonymise run.
competencies = pd.read_csv("../python/output/export/competencies.csv")
```

```python
# One row per (student, competency) verdict -- how many fall in each class.
counts = competencies.groupby("classification").size()
counts
```

```python
# Chart output is cleared before committing to keep the notebook diff-friendly;
# re-run this cell to render it.
ax = counts.plot.bar(title="Competency verdicts across the cohort")
ax.set_xlabel("classification")
ax.set_ylabel("number of (student, competency) rows")
```
